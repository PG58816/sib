### 2.1) dropna.
**Load a version of the dataset that has missing values, then remove every sample containing at least one NaN.**

### 2.2) fillna
**Load the dataset with missing values again and replace the NaNs with a fixed value, the mean, or the median of each feature.**

```python
iris_missing_fixed = read_csv('datasets/iris/iris_missing_data.csv', 
                              sep      = ',', 
                              features = True, 
                              label    = True)

iris_missing_fixed.fillna(0.0)
print('Any NaN left (fillna with 0.0)?', np.isnan(iris_missing_fixed.X).any())

iris_missing_mean = read_csv('datasets/iris/iris_missing_data.csv', 
                             sep      = ',', 
                             features = True, 
                             label    = True)

iris_missing_mean.fillna('mean')
print('Any NaN left (fillna with mean)?', np.isnan(iris_missing_mean.X).any())

iris_missing_median = read_csv('datasets/iris/iris_missing_data.csv',
                               sep      = ',', 
                               features = True, 
                               label    = True)

iris_missing_median.fillna('median')
print('Any NaN left (fillna with median)?', np.isnan(iris_missing_median.X).any())
```

## Exercise 3 — Implementing and testing SelectPercentile

## Exercise 6 — Implementing stratified splitting
### 6.2) Test the stratified_train_test_split function (iris dataset).

```python
from si.model_selection.split import stratified_train_test_split
```

**Load the iris dataset**

```python
iris_split = read_csv('datasets/iris/iris.csv', 
                      sep      = ',', 
                      features = True, 
                      label    = True)

print('Original shape:', iris_split.X.shape)
```

**Split the dataset (20% test).**

```python
train, test = stratified_train_test_split(iris_split, 
                                          test_size    = 0.2, 
                                          random_state = 42)

print('Train shape:', train.X.shape)
print('Test shape: ', test.X.shape)
```

**Check that the class proportions are preserved.**

```python
def class_proportions(dataset):
    classes, counts = np.unique(dataset.y, return_counts = True)
    for c, n in zip(classes, counts):
        print(f'  {c:<16} {n:>3}  ({n / counts.sum():.2%})')

print('Original:')
class_proportions(iris_split)
print('Train:')
class_proportions(train)
print('Test:')
class_proportions(test)
```

## Exercise 7 — Implementing the KNNRegressor with RMSE
### 7.3) Test the KNNRegressor class using the cpu dataset (regression)

```python
from si.models.knn_regressor import KNNRegressor
from si.model_selection.split import train_test_split
```

**Load the cpu dataset.**

```python
cpu = read_csv('datasets/cpu/cpu.csv', 
               sep      = ',', 
               features = True, 
               label    = True)

print('Shape:',    cpu.X.shape)
print('Features:', cpu.features)
print('Label:',    cpu.label)
```

**Split the dataset (20% test).**

```python
cpu_train, cpu_test = train_test_split(cpu, 
                                       test_size    = 0.2, 
                                       random_state = 42)

print('Train shape:', cpu_train.X.shape)
print('Test shape: ', cpu_test.X.shape)
```

**Fit the KNNRegressor and compute the RMSE on the test set.**

```python
knn_reg = KNNRegressor(k = 3)
knn_reg.fit(cpu_train)

score = knn_reg.score(cpu_test)
print(f'RMSE (k = 3): {score:.2f}')
```
