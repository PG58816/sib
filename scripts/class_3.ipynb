```python
import os
from pathlib import Path

import numpy as np

from si.io.csv_file import read_csv
from si.model_selection.split import stratified_train_test_split
from si.models.logistic_regression import LogisticRegression

if Path.cwd().name == 'scripts':
    os.chdir(Path.cwd().parent)

print(os.getcwd())
```

## Testing LogisticRegression
### 1. Use the dataset breast-bin.csv

```python
breast = read_csv('datasets/breast_bin/breast-bin.csv', 
                  sep      = ',', 
                  features = False, 
                  label    = True)

print('Shape:',    breast.X.shape)
print('Features:', breast.features)
print('Label:',    breast.label)
print('Classes:',  breast.get_classes())
```

### 2. Divide the dataset into train and test sets

```python
breast_train, breast_test = stratified_train_test_split(breast, 
                                                        test_size    = 0.2, 
                                                        random_state = 42)

print('Train shape:', breast_train.X.shape)
print('Test shape: ', breast_test.X.shape)
```

### 3. Train the model. Which score do you get? And the cost?

```python
logistic = LogisticRegression(l2_penalty = 1, 
                              alpha      = 0.001, 
                              max_iter   = 1000, 
                              patience   = 5, 
                              scale      = True)
logistic.fit(breast_train)

print('Iterations:', len(logistic.cost_history))
print(f'Score (accuracy): {logistic.score(breast_test):.4f}')
print(f'Cost:             {logistic.cost(breast_test):.4f}')
```
