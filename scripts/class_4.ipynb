## Exercise 1 — Network with 2 inputs, 3 neurons and 1 output (first layer only)

$z = x \cdot W + b$

- $z_1 = 1 \times 0.1 + 2 \times 0.3 + 0.1 = 0.8$
- $z_2 = 1 \times 0.2 + 2 \times (-0.2) + 0 = -0.2$
- $z_3 = 1 \times (-0.1) + 2 \times 0.4 - 0.1 = 0.6$

```python
x = np.array([[1, 2]])                   
W = np.array([[0.1,  0.2, -0.1],
              [0.3, -0.2,  0.4]])         
b = np.array([[0.1, 0, -0.1]])         

z = np.dot(x, W) + b

print(z)
print('Shape:', z.shape)
```

## Exercise 2 — Two layers (2 → 3 → 1) with a random X
Random X with 5 samples and 2 features; weights uniform in [-0.5, 0.5[ and biases set to zero.

```python
np.random.seed(42)

X = np.random.rand(5, 2)                  

# Layer 1: 2 -> 3
W1 = np.random.rand(2, 3) - 0.5           
b1 = np.zeros((1, 3))                   

# Layer 2: 3 -> 1
W2 = np.random.rand(3, 1) - 0.5          
b2 = np.zeros((1, 1))                    

Z1 = np.dot(X, W1) + b1                 
Z2 = np.dot(Z1, W2) + b2                  

print('X: ', X.shape)
print('W1:', W1.shape, ' b1:', b1.shape)
print('Z1:', Z1.shape)
print('W2:', W2.shape, ' b2:', b2.shape)
print('Z2:', Z2.shape)
print()
print('Output:')
print(Z2)
```
