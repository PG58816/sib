## Testing RidgeRegression
### 1. Use the dataset cpu.csv

```python
cpu = read_csv('datasets/cpu/cpu.csv', 
               sep      = ',', 
               features = True, 
               label    = True)

print('Shape:',    cpu.X.shape)
print('Features:', cpu.features)
print('Label:',    cpu.label)
```

### 2. Divide the dataset into train and test sets

```python
cpu_train, cpu_test = train_test_split(cpu, 
                                       test_size    = 0.2, 
                                       random_state = 42)

print('Train shape:', cpu_train.X.shape)
print('Test shape: ', cpu_test.X.shape)
```

### 3. Train the model. Which score do you get? And the cost?

```python
ridge = RidgeRegression(l2_penalty = 1, 
                        alpha      = 0.001, 
                        max_iter   = 1000, 
                        patience   = 5, 
                        scale      = True)
ridge.fit(cpu_train)

print('Iterations:', len(ridge.cost_history))
print(f'Score (MSE): {ridge.score(cpu_test):.2f}')
print(f'Cost:        {ridge.cost(cpu_test):.2f}')
```

## Extra — KNN with Manhattan and Minkowski distances

```python
from functools import partial

from si.model_selection.split import stratified_train_test_split
from si.models.knn_classifier import KNNClassifier
from si.statistics.euclidean_distance import euclidean_distance
from si.statistics.manhattan_distance import manhattan_distance
from si.statistics.minkowski_distance import minkowski_distance
```

**Load and split the iris dataset.**

```python
iris = read_csv('datasets/iris/iris.csv', 
                sep      = ',', 
                features = True, 
                label    = True)

iris_train, iris_test = stratified_train_test_split(iris, 
                                                    test_size    = 0.2, 
                                                    random_state = 42)

print('Train shape:', iris_train.X.shape)
print('Test shape: ', iris_test.X.shape)
```

**Compare the accuracy of the KNNClassifier with each distance.**

```python
distances = {'Euclidean':       euclidean_distance, 
             'Manhattan':       manhattan_distance, 
             'Minkowski (p=3)': partial(minkowski_distance, p = 3)}

for name, distance in distances.items():
    knn = KNNClassifier(k = 3, distance = distance)
    knn.fit(iris_train)
    print(f'{name:<16} accuracy: {knn.score(iris_test):.4f}')
```
